```python
def assign_granularity_2(row):
    if pd.notna(row['Quarter']):
        return "quarterly"
    else:
        return "annual"
        
```

```python
df2["Granularity"] = df2.apply(assign_granularity_2, axis =1)
df2['Granularity'].value_counts()
```
