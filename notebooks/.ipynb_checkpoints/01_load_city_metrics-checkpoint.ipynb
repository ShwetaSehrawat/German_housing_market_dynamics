```python
def assign_granularity(row):
    if pd.notna(row['Month']):
        return "monthly"
    elif pd.notna(row['Quarter']):
        return "quarterly"
    else:
        return "annual"
        
```

```python
df["Granularity"] = df.apply(assign_granularity, axis = 1)
df["Granularity"].value_counts()
```

```python
df["City"] = df["City"].apply(canonical_city)
```

```python
df["City"].unique()
```

```python
df_city = df[df["City"]!= "GREIX_NATIONAL"].copy()
df_national = df[df["City"] == "GREIX_NATIONAL"].copy()
```

```python
print(df_city.shape)
print(df_national.shape)
```

```python
df_city.to_csv("../data/clean/city_metrics_cities.csv", index = False)
df_national.to_csv("../data/clean/city_metrics_national.csv", index = False)
```
